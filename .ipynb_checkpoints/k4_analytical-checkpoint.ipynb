```python
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import networkx as nx
import scipy.linalg

from steinberg import *

plt.rc("text", usetex=False)
plt.rc("font", family = "serif",size=14)
plt.rc("figure",figsize=(12,10))
%config InlineBackend.figure_format = 'retina'
```

Here, we define specific parameter sampling and Laplacian computing functions for the graph $K_4$.

```python
def eq_params_k4():
    """
    Samples equilibrium parameter sets for a 4-state single-cycle graph
    (k12, k23, k34, k41) and (k21, k32, k43, k14), ensuring zero cycle affinity.

    Returns
    -------
    labels_f : list of 4 floats
        Forward rates: [k12, k23, k34, k41]
    labels_r : list of 4 floats
        Reverse rates: [k21, k32, k43, k14]
    """

    # Sample 7 of the 8 parameters freely
    k12 = round_sig(10 ** np.random.uniform(-3, 3), sig_figs=4)
    k23 = round_sig(10 ** np.random.uniform(-3, 3), sig_figs=4)
    k34 = round_sig(10 ** np.random.uniform(-3, 3), sig_figs=4)
    k41 = round_sig(10 ** np.random.uniform(-3, 3), sig_figs=4)

    k21 = round_sig(10 ** np.random.uniform(-3, 3), sig_figs=4)
    k32 = round_sig(10 ** np.random.uniform(-3, 3), sig_figs=4)
    k43 = round_sig(10 ** np.random.uniform(-3, 3), sig_figs=4)

    # Solve for k14 to make the cycle affinity zero
    # A(C) = log[(k12 k23 k34 k41)/(k21 k32 k43 k14)] = 0 → solve for k14
    k14 = (k12 * k23 * k34 * k41) / (k21 * k32 * k43)

    labels_f = [k12, k23, k34, k41]
    labels_r = [k21, k32, k43, k14]

    return labels_f, labels_r

def params_shuffle_k4(params):
    """
    Takes randomly sampled parameters from random_parameters(num_params) and converts to labels_f and labels_r strcuture
    """
    k12 = params[0]
    k23 = params[1]
    k34 = params[2]
    k41 = params[3]

    k21 = params[4]
    k32 = params[5]
    k43 = params[6]
    k14 = params[7]

    labels_f = [k12, k23, k34, k41]
    labels_r = [k21, k32, k43, k14]

    return labels_f, labels_r
    
def lap_k4(labels_f, labels_r):
    """
    Computes the Laplacian matrix for a 4-state single-cycle graph
    (k12, k23, k34, k41) and (k21, k32, k43, k14), ensuring zero cycle affinity.

    Returns
    -------
    labels_f : list of 4 floats
        Forward rates: [k12, k23, k34, k41]
    labels_r : list of 4 floats
        Reverse rates: [k21, k32, k43, k14]
    """
    
    # labels_f = k12, k23, k34, k41
    k12, k23, k34, k41 = labels_f[0], labels_f[1], labels_f[2], labels_f[3]
    # labels_r = k21, k32, k43, k14
    k21, k32, k43, k14 = labels_r[0], labels_r[1], labels_r[2], labels_r[3]
    
    lap = np.array([[-k14-k12, k21, 0, k41], [k12, -k21-k23, k32, 0], [0, k23, -k32-k34, k43], [k14, 0, k34, -k41-k43]],dtype=float)
    
    return lap
```

Next, we perform a few sanity checks to make sure the code is working as it should. First, we check that the area is actually zero when the affinity is zero?

```python
labels_f, labels_r = eq_params_k4()
lap = lap_k4(labels_f, labels_r)
signal = make_observable([1,2,3,4])
is_zero_at_eq(lap,signal)
```

Is B(G) symmetric at equilibrium?

```python
labels_f_eq, labels_r_eq = eq_params_k4()
lap_eq = lap_k4(labels_f_eq, labels_r_eq)
lambdas, w_i, z_i = spectrum_any(lap_eq)
r_i = normalization_factors(w_i, z_i)
Lk_list = projection_matrices(w_i, z_i, r_i)
pi = np.array(pi_dist(lap_eq))
delta_u_star = np.diag(pi)
B_eq = B_matrix(lambdas, Lk_list, delta_u_star)
print(f"B symmetric at equilibrium: {check_symmetric(B_eq)}")
```

Does the numerical area agree with the analytical area?

```python
labels_f, labels_r = eq_params_k4()
lap_eq = lap_k4(labels_f, labels_r)
aff = cycle_affinity(labels_f,labels_r)
print(f"Affinity at equilibrium: {aff:.2e}  (should be ~0)")

signal = make_observable([1,2,3,4])
tau,tau_max = define_tau_range(lap, max_points=500, cap_factor=10.0)
check_analytical_numerical_consistency(signal, lap_eq, tau)
print("PASS: analytical and numerical areas agree at equilibrium")

params_neq = random_parameters(num_params=8)
labels_f, labels_r = params_shuffle_k4(params_neq)
lap_neq = lap_k4(labels_f, labels_r)
aff_neq = cycle_affinity(labels_f, labels_r)
print(f"\nAffinity off equilibrium: {aff_neq:.4f}")

tau,tau_max = define_tau_range(lap, max_points=500, cap_factor=10.0)
a_13, a_31 = asymmetric_autocorrelation(signal, lap_neq, tau_neq)
print(f"tau_max: {tau_neq[-1]:.4f}")
print(f"a_13 at tau_max: {a_13[-1]:.6e}")
print(f"a_31 at tau_max: {a_31[-1]:.6e}")

tau,tau_max = define_tau_range(lap, max_points=500, cap_factor=10.0)
check_analytical_numerical_consistency(signal, lap_neq, tau_neq)
print("PASS: analytical and numerical areas agree off equilibrium")
```

Check that the autocorrelation functions are well behaved

```python
plt.plot(tau, a_13, label='a_13')
plt.plot(tau, a_31, label='a_31')
plt.legend()
plt.show()
```

# Parameter set 2

{(1, 2): 0.008695, (1, 4): 0.3171, (2, 1): 0.002517, (2, 3): 16.022329473190215, (3, 4): 0.06513, (3, 2): 28.02, (4, 1): 0.2901, (4, 3): 0.1177}

    labels_f : list of 4 floats
        Forward rates: [k12, k23, k34, k41]
    labels_r : list of 4 floats
        Reverse rates: [k21, k32, k43, k14]

```python
k12 = 0.008695
k34 = 0.06513
k41 = 0.2901

k21 = 0.002517
k32 = 28.02
k43 = 0.1177
k14 = 0.3171

k23 = (k21 * k32 * k43 * k14) / (k12 * k34 * k41)
print(k23)

labels_f = [k12, k23, k34, k41]
labels_r = [k21, k32, k43, k14]
aff = np.abs(np.log(np.prod(labels_f)/np.prod(labels_r)))
print(aff)
```

```python
print(labels_f)
print(labels_r)
```

```python
lap = lap_k4(labels_f,labels_r)
print(lap)
```

```python
# check that eigenvalues are real and distinct (they need to stay this way)
lambdas, w_i, z_i = spectrum_any(lap)
print(lambdas)
```

```python
signal = [3,5,7,9]
alpha, beta  = 1, 3
```

```python
N = 2000
affinities = np.empty(N, dtype=float)
analytical_areas = np.empty(N)
numerical_areas = np.empty(N)
eigvals_all_r = np.zeros((N, 4), dtype=float)
eigvals_all_c = np.zeros((N, 4), dtype=float)

# perturb edge k32
edge_idx = 1
perturb_edge = labels_r[edge_idx]
perturb_edge
```

```python
for i in tqdm(range(0,N)):
    
    labels_r[edge_idx] = perturb_edge
    
    lap = lap_k4(labels_f, labels_r)
    
    aff = np.abs(np.log(np.prod(labels_f)/np.prod(labels_r)))
    affinities[i] = aff
    
    tau,tau_max = define_tau_range(lap, max_points=500, cap_factor=10.0)
    a_13, a_31 = asymmetric_autocorrelation(signal, lap, tau)

    numerical_areas[i] = numerical_area(a_13, a_31, tau)
    analytical_areas[i] = np.abs(steinberg_analytical_area(signal,lap,alpha=1,beta=3))

    lambdas, w_i, z_i = spectrum_any(lap)
    eigvals_all_r[i, :] = np.sort(lambdas.real)
    eigvals_all_c[i, :] = np.sort(lambdas.imag)
    
    perturb_edge *= 1.01

lambdas, w_i, z_i = spectrum_any(lap)
r_i = normalization_factors(w_i, z_i)
Lk_list = projection_matrices(w_i, z_i, r_i)
pi = np.array(pi_dist(lap))
delta_u_star = np.diag(pi)
B_final = B_matrix(lambdas, Lk_list, delta_u_star)
print("B(G) at final step:")
print(B_final.view())
check_symmetric(B_final, rtol=1e-05, atol=1e-08)
```

```python
# Steinberg curve plot
plt.plot(np.abs(affinities),numerical_areas,linewidth=1, color= 'black',label='analytical')
plt.plot(np.abs(affinities),analytical_areas,linewidth=1,color='blue',label='numerical')
plt.xlabel(r"$\tilde{A}(C)$")
plt.ylabel(r"$\mathcal{I}^{1,3}(G)$")
plt.title("Parameter set 2, Figure 4B")
plt.legend()
plt.savefig("pset2_fig4b.png")
plt.show()
```

```python
# Eigenvalues plots
plt.figure(figsize=(12,5))

# Real part
plt.subplot(1,2,1)
for j in range(4):
    plt.plot(affinities, eigvals_all_r[:, j], label=f"Re(λ{j+1})")
plt.xlabel("Affinity")
plt.ylabel("Re(λ)")
plt.title("Real Parts of Eigenvalues")
plt.legend()

# Imaginary part
plt.subplot(1,2,2)
for j in range(4):
    plt.plot(affinities, eigvals_all_c[:, j], label=f"Im(λ{j+1})")
plt.xlabel("Affinity")
plt.ylabel("Im(λ)")
plt.title("Imaginary Parts of Eigenvalues")
plt.legend()

plt.tight_layout()
#plt.savefig(f"eigenvalue_evolution_edge_{edge_idx}.png", dpi=300)
plt.show()
```
